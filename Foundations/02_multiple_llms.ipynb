### In this section I am experimenting on multiple models from different LLM Providers

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
load_dotenv(override=True)
```

```python
openai_api_key = os.getenv("OPENAI_API_KEY")
anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
gemini_api_key = os.getenv("GOOGLE_API_KEY")
deepseek_api_key = os.getenv("DEEPSEEK_API_KEY")
grok_api_key = os.getenv("GROK_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
openai = OpenAI()
```

```python
request = """
Please come up with one question that can ask different LLm's to check their intelligence, not a mathematical puzzle but a thought provoking question that require intelligence insight, answer only with question, no explanation
"""
messages = [{"role":"user", "content":request}]
response = openai.chat.completions.create(model = "gpt-5.4", messages= messages)
question = response.choices[0].message.content
display(Markdown(response.choices[0].message.content))
```

```python
ANTHROPIC_BASE_URL = "https://api.anthropic.com/v1/"
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1"
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
GROQ_BASE_URL = "https://api.groq.com/openai/v1"
GROK_BASE_URL = "https://api.x.ai/v1"
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
anthropic = OpenAI(api_key=anthropic_api_key, base_url=ANTHROPIC_BASE_URL)
deepseek = OpenAI(api_key=deepseek_api_key, base_url=DEEPSEEK_BASE_URL)
gemini = OpenAI(api_key=gemini_api_key, base_url=GEMINI_BASE_URL)
groq = OpenAI(api_key=groq_api_key, base_url=GROQ_BASE_URL)
grok = OpenAI(api_key=grok_api_key, base_url=GROK_BASE_URL)
openrouter = OpenAI(api_key=openrouter_api_key, base_url=OPENROUTER_BASE_URL)
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
```

```python
answers = {}
def record(model_name, answer):
    if model_name not in answers:
        answers[model_name] = answer
    display(Markdown(answer))
```

```python
def test_llms(model, model_name):
    # max-tokens set for openrouter
    response = model.chat.completions.create(model=model_name, messages=[{"role":"user", "content":"a rondom number between 1 and 30"}])
    response = response.choices[0].message.content
    record(model_name, response)
```

```python
test_llms(openai, "gpt-5.4")
```

```python
test_llms(anthropic, "claude-opus-5")
```

```python
test_llms(gemini,"gemini-3.8-flash")
```

```python
test_llms(deepseek,"deepseek-v4-pro")
```

```python
test_llms(grok,"grok-4.6")
```

```python
test_llms(groq,"qwen/qwen3.6-27b")
```

```python
test_llms(openrouter,"moonshotai/kimi-k2.6")
```
